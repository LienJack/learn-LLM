# 第 5 章：Embedding 与相似度

对应新版 `roadmap.md` 主线入口。

**核心问题**：token id 只是编号，模型怎么学出语义？RAG 的 embedding 又和 token embedding 有什么不同？

**本章关注**：token embedding, retrieval embedding, dot product, cosine similarity, 语义空间。

**轻量实验**：比较 dot product 与 cosine similarity，观察长度和方向的差异。

> 说明：本 notebook 只做最小可运行观察，不做大型训练；如果后续要接入仓库内 API，请先确认 API 已存在。


```python
import math


def dot(a, b):
    return sum(x * y for x, y in zip(a, b, strict=True))

def cosine(a, b):
    return dot(a, b) / (math.sqrt(dot(a, a)) * math.sqrt(dot(b, b)))

query = [1.0, 1.0]
contract_clause = [2.0, 2.0]
unrelated_clause = [-1.0, 0.2]

for name, vector in {"contract": contract_clause, "unrelated": unrelated_clause}.items():
    print(name, {"dot": round(dot(query, vector), 4), "cosine": round(cosine(query, vector), 4)})
```
