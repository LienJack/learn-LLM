对应新版 `roadmap.md` 主线入口。

**核心问题**：模型训练好了，怎么在成本、延迟、吞吐、质量和安全之间做可验证取舍？

**本章关注**：FP16/INT8/INT4, API server, benchmark, observability, release gate, rollback。

**轻量实验**：比较不同部署候选的质量、安全和延迟，决定是否通过发布门禁。

> 说明：本 notebook 只做最小可运行观察，不做大型训练；如果后续要接入仓库内 API，请先确认 API 已存在。


```python
candidates = [
    {"name": "fp16", "eval_score": 0.88, "safety_score": 0.96, "latency_ms": 420},
    {"name": "int8", "eval_score": 0.86, "safety_score": 0.95, "latency_ms": 260},
    {"name": "int4", "eval_score": 0.78, "safety_score": 0.90, "latency_ms": 180},
]

gate = {"min_eval": 0.84, "min_safety": 0.94, "max_latency_ms": 300}
for item in candidates:
    passed = (
        item["eval_score"] >= gate["min_eval"]
        and item["safety_score"] >= gate["min_safety"]
        and item["latency_ms"] <= gate["max_latency_ms"]
    )
    print(item["name"], "release" if passed else "no release")
```
