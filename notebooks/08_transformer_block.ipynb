# 第 8 章：Transformer Block

对应新版 `roadmap.md` 主线入口。

**核心问题**：attention 只是一个模块，完整 block 还缺什么？

**本章关注**：multi-head attention, FFN, residual connection, LayerNorm/RMSNorm, attention complexity。

**轻量实验**：估算 tiny block 参数量和 attention score 的 T^2 成本。

> 说明：本 notebook 只做最小可运行观察，不做大型训练；如果后续要接入仓库内 API，请先确认 API 已存在。


```python
def transformer_block_budget(hidden_size, num_heads, seq_len, ffn_multiplier=4):
    qkv_and_output = 4 * hidden_size * hidden_size
    ffn_hidden = ffn_multiplier * hidden_size
    ffn = hidden_size * ffn_hidden + ffn_hidden * hidden_size
    norm = 2 * hidden_size
    attention_scores = num_heads * seq_len * seq_len
    return {
        "parameters_approx": qkv_and_output + ffn + norm,
        "attention_score_values": attention_scores,
    }

for seq_len in [8, 16, 32]:
    print(seq_len, transformer_block_budget(hidden_size=32, num_heads=4, seq_len=seq_len))
```
