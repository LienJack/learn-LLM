# 第 6 章：从 Bigram 到神经语言模型

对应新版 `roadmap.md` 主线入口。

**核心问题**：最小语言模型如何从统计表走向神经网络？

**本章关注**：bigram count, context window, causal training, generate loop, perplexity。

**轻量实验**：从 tiny corpus 统计 bigram 转移表，并做一个贪心生成。

> 说明：本 notebook 只做最小可运行观察，不做大型训练；如果后续要接入仓库内 API，请先确认 API 已存在。


```python
from collections import Counter, defaultdict

sentences = ["合同 风险 高 <eos>", "合同 条款 风险 <eos>", "医学 科普 谨慎 <eos>"]
counts = defaultdict(Counter)
for sentence in sentences:
    tokens = ["<bos>"] + sentence.split()
    for current_token, next_token in zip(tokens, tokens[1:], strict=False):
        counts[current_token][next_token] += 1

def next_token(token):
    return counts[token].most_common(1)[0][0]

generated = ["<bos>"]
for _ in range(5):
    generated.append(next_token(generated[-1]))
    if generated[-1] == "<eos>":
        break

print("bigram choices:", {k: dict(v) for k, v in counts.items()})
print("generated:", " ".join(generated))
```
