**核心问题**：神经网络到底怎么“学会”？

**本章关注**：参数, loss, 梯度, optimizer, train/eval loop。

**轻量实验**：不用任何框架，手写一个线性模型的一步梯度下降，观察参数如何因 loss 改变。

> 说明：本 notebook 只做最小可运行观察，不做大型训练；如果后续要接入仓库内 API，请先确认 API 已存在。


```python
# y = w*x + b 的一小步训练：目标是让 x=2 时输出接近 5
x, target = 2.0, 5.0
w, b = 0.5, 0.0
lr = 0.1

prediction = w * x + b
loss = (prediction - target) ** 2

grad_prediction = 2 * (prediction - target)
grad_w = grad_prediction * x
grad_b = grad_prediction

w_after = w - lr * grad_w
b_after = b - lr * grad_b
new_prediction = w_after * x + b_after
new_loss = (new_prediction - target) ** 2

print({
    "loss_before": round(loss, 4),
    "loss_after": round(new_loss, 4),
    "w_after": round(w_after, 4),
    "b_after": round(b_after, 4),
})
```
