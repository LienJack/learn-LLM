# 第 9 章：MiniGPT 从零实现

对应新版 `roadmap.md` 主线入口。

**核心问题**：把 tokenizer、embedding、Transformer block 和 LM head 组合起来，是不是就得到 GPT？

**本章关注**：decoder-only GPT, causal LM, train/generate loop, checkpoint, perplexity。

**轻量实验**：用一个 tiny 配置对象检查 MiniGPT 的组件是否齐全，并生成可测试的接口清单。

> 说明：本 notebook 只做最小可运行观察，不做大型训练；如果后续要接入仓库内 API，请先确认 API 已存在。


```python
mini_gpt_config = {
    "vocab_size": 128,
    "context_length": 16,
    "hidden_size": 32,
    "num_layers": 2,
    "num_heads": 4,
}
required_components = [
    "token_embedding",
    "position_or_rope",
    "transformer_blocks",
    "lm_head",
    "generate",
    "save_load",
]

checks = {
    "head_dim_is_integer": mini_gpt_config["hidden_size"] % mini_gpt_config["num_heads"] == 0,
    "context_is_small": mini_gpt_config["context_length"] <= 32,
    "has_required_components": len(required_components) == 6,
}

print("config:", mini_gpt_config)
print("checks:", checks)
assert all(checks.values())
```
