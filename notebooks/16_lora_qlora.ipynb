# 第 16 章：LoRA / QLoRA 参数高效微调

对应新版 `roadmap.md` 主线入口。

**核心问题**：全量微调太贵，能不能只训练少量参数？

**本章关注**：PEFT, rank/alpha, target_modules, adapter 保存加载, QLoRA, 显存优化。

**轻量实验**：估算 LoRA adapter 参数量，比较不同 rank 的训练规模。

> 说明：本 notebook 只做最小可运行观察，不做大型训练；如果后续要接入仓库内 API，请先确认 API 已存在。


```python
def lora_params(in_features, out_features, rank):
    return rank * (in_features + out_features)

full_linear = 4096 * 4096
for rank in [4, 8, 16]:
    adapter = lora_params(4096, 4096, rank)
    print({
        "rank": rank,
        "adapter_params": adapter,
        "percent_of_full_linear": round(adapter / full_linear * 100, 4),
    })
```
