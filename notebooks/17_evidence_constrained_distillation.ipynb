# 第 17 章：证据约束蒸馏

对应新版 `roadmap.md` 主线入口。

**核心问题**：大模型效果好但太贵，怎么把可验证行为迁移给小模型？

**本章关注**：teacher/student, evidence-grounded generation, unsupported answer reject, 固定 eval 对比。

**轻量实验**：过滤没有证据引用的 teacher 样本，只留下可蒸馏样本。

> 说明：本 notebook 只做最小可运行观察，不做大型训练；如果后续要接入仓库内 API，请先确认 API 已存在。


```python
teacher_samples = [
    {"answer": "需要人工复核", "evidence": ["law-001#s1"]},
    {"answer": "一定合法", "evidence": []},
    {"answer": "unknown", "evidence": ["med-001#s2"]},
]

filtered = [
    sample
    for sample in teacher_samples
    if sample["evidence"] and sample["answer"] != "一定合法"
]
rejected = [sample for sample in teacher_samples if sample not in filtered]

print("kept_for_student_training:", filtered)
print("rejected:", rejected)
```
