# 第 3 章：概率、Softmax、交叉熵与 next-token

对应新版 `roadmap.md` 主线入口。

**核心问题**：如果模型不是分类图片，而是继续写一句话，它该怎么训练？

**本章关注**：next-token prediction, teacher forcing, logits/Softmax, cross entropy, perplexity。

**轻量实验**：手算一个三分类 logits 的 softmax、交叉熵和困惑度。

> 说明：本 notebook 只做最小可运行观察，不做大型训练；如果后续要接入仓库内 API，请先确认 API 已存在。


```python
import math

vocab = ["我", "喜欢", "模型"]
logits = [0.2, 1.7, 0.6]
target_index = 1

max_logit = max(logits)
exp_values = [math.exp(x - max_logit) for x in logits]
probs = [x / sum(exp_values) for x in exp_values]
loss = -math.log(probs[target_index])
perplexity = math.exp(loss)

print("probabilities:", dict(zip(vocab, [round(p, 4) for p in probs], strict=True)))
print("cross_entropy:", round(loss, 4))
print("perplexity:", round(perplexity, 4))
```
