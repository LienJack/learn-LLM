**核心问题**：在开始 RAG、SFT、LoRA 之前，怎么先知道模型有没有变好？

**本章关注**：eval set, metrics, JSON schema validation, refusal accuracy, failure cases。

**轻量实验**：对两个 toy 输出做 JSON 字段校验和格式准确率统计。

> 说明：本 notebook 只做最小可运行观察，不做大型训练；如果后续要接入仓库内 API，请先确认 API 已存在。


```python
required_fields = {"answer", "citations", "needs_human_review"}
outputs = [
    {"answer": "unknown", "citations": [], "needs_human_review": True},
    {"answer": "合同风险较高", "needs_human_review": True},
]

results = []
for item in outputs:
    missing = required_fields - set(item)
    results.append({"valid_format": not missing, "missing": sorted(missing)})

format_accuracy = sum(r["valid_format"] for r in results) / len(results)
print("results:", results)
print("format_accuracy:", format_accuracy)
```
