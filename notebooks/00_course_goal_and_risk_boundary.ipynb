**轻量实验**：用一个内存中的实验配置生成可追溯日志，并检查高风险领域是否必须进入人工复核。

> 说明：本 notebook 只做最小可运行观察，不做大型训练；如果后续要接入仓库内 API，请先确认 API 已存在。


```python
import json
from datetime import datetime, timezone

experiment = {
    "chapter": 0,
    "seed": 42,
    "device": "cpu",
    "domain": "medical",
    "task": "learning-boundary-check",
    "data_manifest": [
        {"source_id": "toy-001", "license": "teaching-only", "contains_pii": False},
    ],
}

risk_policy = {
    "legal": "must cite source and avoid legal advice",
    "medical": "must avoid diagnosis and escalate red flags",
}

log_record = {
    "created_at": datetime.now(timezone.utc).isoformat(),
    "experiment": experiment,
    "risk_boundary": risk_policy.get(experiment["domain"], "general caution"),
    "needs_human_review": experiment["domain"] in {"legal", "medical"},
}

print(json.dumps(log_record, ensure_ascii=False, indent=2))
```
