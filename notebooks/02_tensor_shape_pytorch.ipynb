**核心问题**：为什么 LLM 学习里 shape 比公式更容易卡人？

**本章关注**：[B,T], [B,T,C], broadcasting, matrix multiplication, device 与 dtype。

**轻量实验**：用嵌套 list 模拟 token ids、embedding 和 logits 的关键 shape。

> 说明：本 notebook 只做最小可运行观察，不做大型训练；如果后续要接入仓库内 API，请先确认 API 已存在。


```python
def shape(x):
    dims = []
    while isinstance(x, list):
        dims.append(len(x))
        x = x[0] if x else None
    return dims

input_ids = [[1, 2, 3], [2, 3, 0]]  # [B, T]
embedding_table = {
    0: [0.0, 0.0],
    1: [0.2, -0.1],
    2: [0.4, 0.3],
    3: [-0.5, 0.8],
}
embeddings = [[embedding_table[token] for token in row] for row in input_ids]  # [B, T, C]
logits = [[[sum(vec), -sum(vec)] for vec in row] for row in embeddings]  # [B, T, V=2]

print("input_ids shape:", shape(input_ids))
print("embeddings shape:", shape(embeddings))
print("logits shape:", shape(logits))
```
