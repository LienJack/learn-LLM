**核心问题**：如何做一个谨慎、安全、可评测的医学科普助手？

**本章关注**：医学科普问答, 红旗症状, 指南 RAG, 药物禁忌, 不替代医生诊断。

**轻量实验**：识别红旗症状并生成必须人工/临床复核的输出骨架。

> 说明：本 notebook 只做最小可运行观察，不做大型训练；如果后续要接入仓库内 API，请先确认 API 已存在。


```python
question = "胸痛并且呼吸困难，可以先观察吗？"
red_flags = [term for term in ["胸痛", "呼吸困难", "意识不清", "大出血"] if term in question]

output = {
    "summary": "存在需要及时就医评估的危险信号。",
    "possible_explanations": [],
    "red_flags": red_flags,
    "seek_care_level": "urgent" if red_flags else "unknown",
    "self_care_general_info": "不要用本系统替代医生诊断。",
    "medication_warning": "未评估个体用药禁忌。",
    "evidence": [],
    "uncertainty": "缺少年龄、病史、持续时间等信息。",
    "not_diagnosis": True,
    "needs_clinician_review": True,
}

print(output)
assert output["not_diagnosis"] is True
assert output["needs_clinician_review"] is True
```
