**核心问题**：领域模型的能力主要来自模型，还是任务定义和数据？

**本章关注**：任务边界, 许可矩阵, 脱敏, 去重, train/eval leakage, data card。

**轻量实验**：把同一条领域样本拆成 SFT、RAG、distill、eval 四种用途，并检查许可和泄漏风险。

> 说明：本 notebook 只做最小可运行观察，不做大型训练；如果后续要接入仓库内 API，请先确认 API 已存在。


```python
sample = {
    "source_id": "contract-toy-001",
    "text": "违约金不得超过实际损失的30%（教学示例，不作为法律依据）。",
    "license": "teaching-only",
    "contains_pii": False,
}

artifacts = {
    "sft": {"prompt": "解释条款风险", "answer_from": sample["source_id"]},
    "rag": {"chunk": sample["text"], "source_id": sample["source_id"]},
    "distill": {"teacher_must_cite": sample["source_id"]},
    "eval": {"question": "是否能直接给法律意见？", "expected": "不能"},
}

quality_checks = {
    "license_known": bool(sample["license"]),
    "pii_removed": not sample["contains_pii"],
    "eval_source_traceable": bool(sample["source_id"]),
}

print("artifacts:", artifacts)
print("quality_checks:", quality_checks)
```
