# 第 7 章：Attention 手算与实现

对应新版 `roadmap.md` 主线入口。

**核心问题**：一句话中每个 token 怎么决定自己应该看谁？

**本章关注**：Q/K/V, scaled dot-product attention, causal mask, attention weights, [B,H,T,T]。

**轻量实验**：用三个二维 token 手写 causal attention，确认未来 token 权重为 0。

> 说明：本 notebook 只做最小可运行观察，不做大型训练；如果后续要接入仓库内 API，请先确认 API 已存在。


```python
import math


def softmax(values):
    m = max(values)
    exp_values = [math.exp(v - m) for v in values]
    total = sum(exp_values)
    return [v / total for v in exp_values]

tokens = [[1.0, 0.0], [0.5, 0.5], [0.0, 1.0]]
scale = math.sqrt(2)
all_weights = []

for i, query in enumerate(tokens):
    scores = []
    for j, key in enumerate(tokens):
        score = sum(q * k for q, k in zip(query, key, strict=True)) / scale
        scores.append(float("-inf") if j > i else score)
    weights = softmax(scores)
    all_weights.append([round(w, 4) for w in weights])

print("causal attention weights:")
for row in all_weights:
    print(row)
assert all_weights[0][1] == 0 and all_weights[0][2] == 0
```
